#### Checking for Empty values or empty lists

```python
empty_list = pd.DataFrame({
    '[] count':(df == '[]').sum(),
    'percentage':((df == '[]').sum()/len(df)).round(2)
})
print(empty_list)
```

#### Dropping Unusefull columns or columns with many Empty lists

```python
df = df.drop(columns=['url','phone','menu_item'])
df.head()
```

```python
df = df.dropna(subset = ['rest_type', 'location', 'approx_cost(for two people)', 'cuisines'])
```

```python
df.isnull().sum()
```

#### Handling Duplicate

```python
df[df.duplicated()]
```

```python
df = df.drop_duplicates()
```

#### Simplifying and cleaning the columns

```python
df['rate'].unique()
```

```python
(df['rate'].isin(['NEW', '-'])).sum()
```

```python
df['rate'].value_counts()[['NEW', '-']]
```

```python
df['rate'] = df['rate'].replace(['NEW', '-'], np.nan)
```

```python
df['rate'].unique()
```

```python
df['rate'] = df['rate'].str.strip()
df['rate'] = df['rate'].str.replace('/5','', regex=False)
df['rate'] = pd.to_numeric(df['rate'], errors = 'coerce')
#"coerce means If you cannot convert a value to a number, don't throw an error. Instead, replace it with NaN."
df['rate']
```

```python
df['approx_cost(for two people)'] = (
    df['approx_cost(for two people)']
    .str.replace(',', '', regex=False)
)
df['approx_cost(for two people)'] = pd.to_numeric(df['approx_cost(for two people)'], errors = 'coerce' )
```

```python
cols = ['address',
        'name',
        'location',
        'rest_type',
        'dish_liked',
        'cuisines',
        'reviews_list',
        'listed_in(type)',
        'listed_in(city)']
df[cols] = df[cols].apply(lambda col: col.str.strip())
```

```python
df['rest_type'] = df['rest_type'].str.replace(
    'Irani Cafee',
    'Irani Cafe',
    regex=False
)
```

```python
df.info()
```

#### Checking for empty values , or characters 

```python
df.isin(['',' ', "", " "]).sum()
```

```python
df[df['rest_type'].str.fullmatch(r'[^A-Za-z0-9]+', na=False)]
# r'[^A-Za-z0-9]' = any character that is not a letter or number. and + means one or more like ###,//, $$ etc 
```

#### Mapping Yes and No to 1 and 0

```python
df['online_order'] = df['online_order'].map({'Yes': 1, 'No': 0})
df['book_table'] = df['book_table'].map({'Yes': 1, 'No': 0})

df[['online_order', 'book_table']].head()
```
